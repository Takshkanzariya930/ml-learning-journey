## Feature Preparation & Preprocessing Strategy

### Separating Features and Target

````python
x = df.drop(columns="default_payment_next_month")
y = df["default_payment_next_month"]

df.to_csv("cleaned_data.csv")
````

#### Why target must be separated early ?
- We must separate target from feature to prevent target leakage and to prevent targets from influencing model in training phase. Model should predict target not observe them.

#### What could go wrong if it isn’t ?
- Model can be influenced by target and will not predict new instances accurately. or will not work in real life on real data which is our ultimate goal.

### Building the Preprocessing Pipeline

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

preprocessor = ColumnTransformer(transformers=[ ("num", StandardScaler(), numeric_feature) ], remainder="passthrough")
````

#### What ColumnTransformer does ?
- The ColumnTransformer in sklearn is a powerful tool for applying different preprocessing transformations to different columns of a dataset in a streamlined and efficient way.

#### Why this is safer than manual preprocessing ?
- Because it prevents data leakage and it apples transformation to specific columns instead of all the columns of dataset.

## Baseline Model — Logistic Regression

### Train/Test Split

````python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
````

````python
figure, axes= plt.subplots(1,2, figsize=(12, 5))

sns.countplot(x=y_train, ax=axes[0])
axes[0].set_title("y_train")
sns.countplot(x=y_test, ax=axes[1])
axes[1].set_title("y_test")
````

````python
y_train.value_counts(normalize=True), y_test.value_cxounts(normalize=True) 
````

#### Why we split ?
- We split the dataset so that we can test our model on completely unseen data to accurately score or check performance of our model.

#### Why stratify=y matters here ?
- Stratify=y ensures that the proportion of each class in the training and test sets matches the proportion in the original dataset.  This is crucial for classification problems, especially with imbalanced datasets.

#### What could go wrong without stratification ?
- Training or testing set might not have enough variations of distinct classes to train or evaluate model so that it predictions are correct on real data and not biased towards one class.

### Building FULL Pipeline

````python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(max_iter=1000, class_weight="balanced"))
    ]
)
````

#### Why preprocessing is inside the pipeline ? 
- Preprocessing is integrated into the pipeline to ensure consistency, prevent data leakage, and enable reproducible, automated workflows.

#### Why class_weight="balanced" is used ?
- class_weight="balanced" is used to automatically adjust the weights of classes in imbalanced datasets, ensuring that the model pays more attention to the minority class during training.  This helps prevent the model from being biased toward the majority class, which can lead to poor performance on the minority class. 

#### What bias–variance effect this has ?
- class+weight="balanced" reduces bias of model toward majority class by adjusting the loss function to penalize misclassifications of the minority class more heavily.
- This can also increase variance as model will give more importance to minority class and it will overfit to minority class noise.

### Cross-Validation on TRAINING SET ONLY

````python
from sklearn.model_selection import cross_validate

scoring = ["accuracy", "precision", "recall", "roc_auc"]

cv_result = cross_validate(model, x_train, y_train, cv=5, scoring=scoring)

for metric in scoring:
    
    print(f"{metric} : ", cv_result[f"test_{metric}"].mean())
    
cv_result
````

#### Accuracy :-
- Accuracy tells you how many predictions did model get right out of all the predictions. In our uses case accuracy should be around 0.80 or more because every misclassification leads to loss to credit provider.

#### Precision :- 
- Precision focus on the quality of the model’s "default" predictions. It tells us how many of the "default" predictions were actually correct out of all the predictions that were classified as "default". In our case precision needs to be around 0.90 or more because every misclassified "default" leads to inconvenience to customer.

#### Recall :-
- Recall measures how good the model is at predicting "default". It shows the proportion of "default" detected out of all the actual "default" instances. In our case recall needs to be around 0.80 or more because every misclassification of "default" leads to loss to credit provider. 

#### ROC-AUC :- 
- ROC-AUC represents how well the model is able to separate "default" form "Non-default" instant. We will use it to compare model and it should be 0.8 or above but should not reach above 0.9 because it means that their is data leakage.
 
#### Which metric matters most for this business problem ?
- In our case recall matters most because not identifying "default" is direct financial loss. But precision is also considered to limit false alarm or misclassification of "Non-default" as "default" which will lead to decrease in customer's satisfaction level. while ROC–AUC is used to evaluate overall class separability. Accuracy is not a reliable metric due to class imbalance.

### Fitting on Training Data, Evaluating on Test Set

````python
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score, confusion_matrix

model.fit(x_train, y_train)

y_pred = model.predict(x_test)
y_prob = model.predict_proba(x_test)[: ,1]

print("Accuracy : ", accuracy_score(y_test, y_pred))
print("Precision : ", precision_score(y_test, y_pred))
print("Recall : ", recall_score(y_test, y_pred))
print("ROC-AUC : ", roc_auc_score(y_test,y_prob))
print("Confusion Matrix : \n", confusion_matrix(y_test, y_pred))
````

#### Is the model biased toward majority class ?
- By calculating recall for "default" and "Non-default" we can say that if it recall of one class is very high compare to other then model is biased towards class with high recall. recall for "default" (positive) is 0.63 (TP / (TP + FN)) and recall for "Non-default" id 0.70 (TN / (TN + FP)). As their is no large difference between both recalls model is not that biased.

#### Is recall acceptable for defaulters ?
- Currently recall is 0.63 which is solid for baseline but not acceptable for "default" as we will need more recall to minimize loss caused due to "default".

#### What errors are most costly here ?
- False Negatives or misclassified "default" as "Non-default" are very costly as they will lead to direct loss to credit provider.

#### Is this baseline strong enough to beat ?
- The baseline Logistic Regression model achieves a recall of ~63% for defaulters, indicating that the majority of risky customers are correctly identified. However, precision is relatively low (~38%), meaning a significant number of non-defaulters are flagged as risky. This trade-off is expected in a recall-focused setup for credit risk, where missing defaulters is more costly than issuing false alarms. The ROC–AUC of ~0.72 confirms that the model has reasonable class-separation ability, making it a suitable baseline to improve upon.

## Hyperparameter Tuning and Threshold Optimization

### Choosing the Model to Tune

#### Model selected for tuning
- Random Forest is selected due to it's high ROC-AUC, which indicates better class separability. Our goal is to improve *recall* of this model while maintaining *precision*.

````python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "classifier__n_estimators" : [200, 400],
    "classifier__max_depth" : [None, 6, 10],
    "classifier__min_samples_split" : [2, 10]
}
````

#### Why these ?
- **max_depth** :-
    - "max_depth" controls how deep tree grows. if tree grows to deeps it overfits the data as separability of model is high but recall is low which indicates model is overfitting the data. So we reduce "max_depth" thus decreasing overfitting (variance) and improving recall.

- **min_sample_split** :-
    - "min_sample_split" controls how many sample requires before tree makes a split. if number is small tree splits aggressively and learns rare patters which again leads to overfitting. So we increase "min_sample_split" thus forcing tree to learn broader and more stable pattern which decreases overfitting (variance) and improve recall.
    
- **n_estimator** :-    
    - "n_estimator" controls number of trees in the forest. more trees lead to stable probability outputs and reduce randomness thus increasing stability of model. it will help us lather in threshold optimization. 

### Tuning Using ROC-AUC 

````python
rf_grid = GridSearchCV(
    rf_model,
    param_grid,
    cv=5,
    scoring="roc_auc",
    n_jobs=-1
)

rf_grid.fit(x_train, y_train)

rf_grid.best_params_, rf_grid.best_score_
````

#### Did ROC-AUC improve ?
- Yes, it went from 0.76 to 0.78.

#### Was the improvement meaningful or marginal?
- Improvement was marginal but meaningful as this slight increase in ROC-AUC suggest that our model is more stable, it's probability estimates are reliable and it has better class separability.  
- Further improvement will come from Threshold tuning.

### Evaluating Tuned Model on Test Set

````python
best_rf = rf_grid.best_estimator_

y_pred = best_rf.predict(x_test)
y_prob = best_rf.predict_proba(x_test)[: ,1]

print("Accuracy : ", accuracy_score(y_test, y_pred))
print("Precision : ", precision_score(y_test, y_pred))
print("Recall : ", recall_score(y_test, y_pred))
print("ROC-AUC : ", roc_auc_score(y_test,y_prob))
print("Confusion Matrix : \n", confusion_matrix(y_test, y_pred))
````

#### Comparing baseline RF vs tuned RF:
- **Accuracy** :-
    - Accuracy decreased from base model but it does not much matter as our data is imbalanced.
- **Precision** :-  
    - Precision decreased from base model but in our case we can accept that as long as it not too low and recall is high.
- **Recall** :- 
    - Recall improved significantly which directly comes in our interest because lower recall leads to direct loss.
- **ROC-AUC** :-
    - ROC-AUC also increased though marginal it has improved.

### Threshold Optimization

````python

thresholds = [0.3, 0.35, 0.4, 0.45, 0.5]

for t in thresholds:
    
    preds = (y_prob >= t).astype(int)
    
    print(
        f"Threshold {t:.2f} | "
        f"Precision: {precision_score(y_test, preds):.3f} | "
        f"Recall: {recall_score(y_test, preds):.3f}"
    )
    
print("\n\nAt Threshold o.35 :- \n")
    
preds = (y_prob >= 0.35).astype(int)
    
print("Accuracy : ", accuracy_score(y_test, preds))
print("Precision : ", precision_score(y_test, preds))
print("Recall : ", recall_score(y_test, preds))
print("ROC-AUC : ", roc_auc_score(y_test,y_prob))
print("Confusion Matrix : \n", confusion_matrix(y_test, preds))
````

````python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, preds)

plot = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Non-defaulters", "defaulters"])

ax = plt.subplot()

plot.plot(ax=ax)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Random Forest (Tuned) | Confusion Matrix")
````

#### Why 0.35 is better trade-off ?
- In a credit default prediction context, false negatives (missed defaulters) are significantly more costly than false positives (customers incorrectly flagged as risky). At a threshold of 0.35, the model achieves a recall of approximately 78%, meaning most defaulters are identified, while maintaining a non-trivial precision of 34%. This threshold provides a reasonable trade-off between risk mitigation and operational cost, and is preferred over more conservative thresholds that miss a large fraction of defaulters.
