### Typecasting
- Convert data from one type to another using built-in functions.

```python
# str to int
x = int("10") # 10

# int to str
y = str(25) # "25"

# float to int
z = int(3.9) # 3 (truncates, not rounds)

# list from string
lst = list("abc") # ['a', 'b', 'c']
```
