## What is Tensor?

- Tensor is a data structure used to store numbers. Can be of N-dimensions. 

- 0D tensor
    - 0D tensor is nothing but a 0D array or scaler value.
    - Example :- 5 -> 0D tensor, Rank 0

- 1D tensor
    - 1D tensor is nothing but vector or an 1D array and number of elements inside this 1D tensor determines dimension of the vector.
    - Example :- [1,2,3] -> 1D tensor and 3D vector, Rank 1

- 2D tensor 
    - 2D tensor is nothing but collection of 1D tensor also knows as matrix.
    - Example :- [[1,2,3], [4,5,6]] -> 2D tensor or matrix, Rank 2

- 3D tensor
    - 3D tensor is collection of 2D tensor.
    - Example :- In the domain of NLP when a collection of sentence is represented using 3D tensor and one sentence is represented by matrix which shows which words are present in the sentence. 

- 4D tensor
    - 4D tensor is collection of 3D tensor.
    - Example :- In the domain of CV a collection of images are represented by 4D tensor while one image is represented in 3D tensor.

### Why are tensors useful for deep learning?
- As we show above tensor can be used to efficiently represent text, image, video and other high-dimensional data and this provides computer a mean to understand and make computation on this data thus they are very useful in deep learning because in deep leaning we are dealing with high-dimensional data. 

### Why are tensors better suited for GPU computation than Python lists?
- Tensors are suitable for GPU computation because they are homogenous, contiguously stored, optimized for parallel mathematical operation and are nicely integrated though libraries like Pytorch and Tensorflow to use multiple cores of GPU to complete computation task.

### When would random initialization be useful?
- Random initialization of tensors is primarily useful for initializing the weights of neural networks to break symmetry and ensure that neurons learn different features during training.

### Why are neural network weights not initialized to all zeros?
- If weights are initialized to the same value (such as zero or one), all neurons in a layer will compute identical outputs and receive identical gradient updates, effectively reducing the network's capacity to learn complex patterns.

### Why does deep learning usually use floating point tensors?
- Because floating point tensors provide necessary precision and range to handle the fractional values and extreme magnitudes inherent in neural network computations

### What happens if integer tensors are used during gradient computation?
- If integers are used small gradient changes would be rounded up to zero and learning will be halted.

## Basic Tensor Operations

```python
x = torch.tensor([[1, 2],
                  [3, 4]])

y = torch.tensor([[5, 6], 
                  [7, 8]])

x + y
```

```python
x * y
```

### Is * matrix multiplication here?
- No

### What operation is actually happening?
- multiplication of numbers which are in same position.

## Matrix Multiplication

```python
x = torch.tensor([[1, 2],
                  [3, 4]])

y = torch.tensor([[5, 6], 
                  [7, 8]])

torch.matmul(x, y)
```

### Why is matrix multiplication fundamental in neural networks?
- Matrix multiplication is fundamental in neural networks because it provides means to perform linear transformations across layers. 

### What shape rules must hold?
- No. of columns of the first matrix should be equal to No. of rows of the second matrix for matrix multiplication. 

## Tensor Indexing

```python
x = torch.tensor([
    [[1, 2],
    [3, 4]],
    
    [[5, 6],
    [7, 8]]
])
```

```python
print(x[0])
print(x[0,1])
print(x[:, 1])
```

- GPUs are useful for deep learning because their architecture is optimized for parallel processing, allowing them to perform thousands of simultaneous matrix and tensor operations that are central to neural network training.
