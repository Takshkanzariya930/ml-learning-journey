## Cross Validation

```python
from sklearn.datasets import make_classification

x, y = make_classification(n_samples=100, n_features=3, n_informative=3, n_redundant=0, n_classes=2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

models = { "Logistic Regression" : LogisticRegression(max_iter=1000),
            "KNN" : KNeighborsClassifier(n_neighbors=5),
            "Decision Tree" : DecisionTreeClassifier(random_state=42),
            "Random Forest" : RandomForestClassifier(n_estimators=100, random_state=42) }
```

```python
import numpy as np
from sklearn.model_selection import cross_val_score

for model in models:
    scores = cross_val_score(models[model], x, y, cv=5, scoring="accuracy")
    scores = np.array(scores)
    print(model , np.mean(scores, axis=0), np.std(scores, axis=0))
```

### Why Mean score matters ?

- We take mean of score in k fold cross validation to better estimate performance of model on unseen data, It serves as average accuracy of model 

### Why std matters ?

- We use std to quantify variability and uncertainty of a model across different folds of data. As std gets lower consistency of model increases and visa-versa.

## Model Comparison

- We select model on main Four merits as we cannot trust one merit completely.

- **Average Accuracy** : We want our Average cross-validation score to be high. Higher the score better model's predictive performance. But never trust training accuracy or single train-test split score.

- **Stability** : Stability is quantified by std in cross-validation score. we want std to be low. lower std means model is more consistent. A slightly less accurate but more stable model is often better in production.

- **Bias-Variance Interpretation** : 

    | Symptom             | Meaning                     |
    | ------------------- | --------------------------- |
    | Low mean, low std   | High bias (underfitting)    |
    | High mean, high std | High variance (overfitting) |
    | High mean, low std  | Sweet spot (ideal)          |
    
- **Practical Constrains** : This is also a deciding factor. field you are working in or your need. Example in medical field or model which predicts disease their you want maximum accuracy because there life is at risk. In some places you might prefer stability over accuracy.

### Why is cross-validation more reliable than a single split ?

- Cross-validation is more reliable than a single train-test split because it provides a more robust and less biased estimate of a model's performance by averaging results across multiple data splits.

### What does a high standard deviation across folds indicate ?

- Model is inconsistent and model's prediction varies significantly between different folds. This also shows model is sensitive to specific data points.

### Why might you choose a slightly less accurate but more stable mode ?

- Choosing a slightly less accurate but more stable mode can be advantageous in scenarios where consistent performance is more critical than achieving the highest possible accuracy.
-  We will want to see prediction at least close to the actual value rather then completely off values which might hurt where it is being used.
