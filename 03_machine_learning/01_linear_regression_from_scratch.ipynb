## Linear Regression form scratch

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])
```

We start with random or zero values because we need some values to compare against which gives direction to move on.

```python
w = 0
learning_rate = 0.01
epochs = 100
```

In every iteration prediction is made in "y_pred" then loss is calculated by mean squared error in "loss" now gradient is calculated here differentiation of "(y-y_pred)**2" is done manually and feed to mean function to find slop/gradient then weight is corrected based in gradient every step value is "gradient * leaning_rate" which is then subtracted from weight and thus we get correct weight to data and thus accurate prediction.

```python
for epoch in range(epochs):
    
    y_pred = x * w
    loss = np.mean((y - y_pred) ** 2)
    
    gradient = -2 * np.mean(x * (y - y_pred))
    w = w - (gradient * learning_rate)
    
    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Loss {loss:.4f}, Weight {w:.4f}, Gradient {gradient: .4f}")
```

```python
x_test = np.array([7,8,9])

print(x_test * w)
```

prediction on new values of 'x' is correct thus our model is trained.
